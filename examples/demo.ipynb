```python
%%slide summary
# The same slide, `midnight` theme

- Only the theme changed; the cell body is untouched.
- Built-in themes: `amber` (default), `slate`, `midnight`.
```

### Your own palette and your own font

`font_family` and `font_url` accept any font: a web font, or a locally installed one (pass `font_url=none` to skip the download).

```python
%slide_theme reset

import notebook_slide as ns

ns.configure(
    accent="#0f766e",
    background="#f7fbfa",
    heading="#0b3b36",
    subheading="#3f6b66",
    border="#c3ded9",
    quote_background="#e3f2ef",
    code_background="#e3f2ef",
    code_foreground="#0b3b36",
    # Any font works: a web font ...
    font_family="'IBM Plex Sans', 'Segoe UI', sans-serif",
    font_url="https://fonts.googleapis.com/css2?family=IBM+Plex+Sans&display=swap",
    # ... or a locally installed one, with font_url=None.
)
```

```python
%%slide preview
# A custom palette and font

- The palette and the typeface are just theme fields.
- `font_url=none` renders with locally installed fonts only.
```

### Overriding a single slide

Any theme field works as a `key=value` on the magic line, and applies to that slide alone.

```python
%%slide summary color=#a33a2b label="One-off styling" background=#fdf3f1 border=#e8c7c0
# This slide only

- `color`, `label` and any theme field can be set per slide.
- The global theme is untouched — the next slide goes back to it.
```

## Custom and translated slide types

`register_slide_type` adds new types or replaces the built-in ones, which is how
you translate the badges. `register_alias` adds another name for the magic.

```python
ns.register_slide_type("demo", "Live demo", "#aa00aa")
ns.register_slide_type("warning", "Careful", "#b45309", background="#fff7ed", border="#f0cfa0")
ns.register_slide_type("resumen", "Recapitulación", "#287681")
ns.register_slide_type("seccion", "Sección", "#4a3f8f", layout="title")
ns.register_alias("diapositiva")
```

```python
%%slide warning
# Careful with `head -n 1000`

- A truncated sample can hide the very rows that break your schema.
```

```python
%%diapositiva seccion
# Parte dos — Almacenamiento columnar
```

```python
%%diapositiva resumen
# Lo que llevamos

- Las etiquetas se traducen registrando el tipo de diapositiva.
- El alias `%%diapositiva` es el mismo magic con otro nombre.
```
